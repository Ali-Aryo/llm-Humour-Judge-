```python
from huggingface_hub import hf_hub_download, list_repo_files
import pandas as pd
```

```python
files = list_repo_files("yguooo/newyorker_caption_ranking", repo_type="dataset")
print(len(files))

ranking_files = [f for f in files if f.startswith("ranking")]
print(ranking_files[:20])
```

```python
path = hf_hub_download(
    repo_id="yguooo/newyorker_caption_ranking",
    filename="ranking/source/510.csv",
    repo_type="dataset",
)
print(path)
```

```python
df = pd.read_csv(path)
print(df.shape)
```

```python
print(df.columns.tolist())
df.head(10)
```

```python
df[["mean", "votes"]].describe()
```

```python
df.tail(10)
```

```python
keep = df[df["votes"] >= 10]
print(len(keep), "captions have at least 10 votes")
```

```python
keep = keep.sort_values("mean", ascending=False).reset_index(drop=True)
n = len(keep)

top = keep.head(5)
middle = keep.iloc[n//2 - 2 : n//2 + 3]
bottom = keep.tail(5)

sample = pd.concat([
    top.assign(level="top"),
    middle.assign(level="middle"),
    bottom.assign(level="bottom"),
])
sample[["level", "caption", "mean", "votes"]]
```

```python
sample.to_csv("contest510_sample.csv", index=False)
```
